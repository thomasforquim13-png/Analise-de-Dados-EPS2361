```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Criando um pequeno conjunto de dados de exemplo para teste
dados = pd.DataFrame({
    'Atividade': ['Planejamento', 'Limpeza', 'Análise', 'Relatório'],
    'Horas': [5, 10, 15, 8]
})

print("Bibliotecas carregadas com sucesso!")
print(dados)

# Criando um gráfico simples
sns.barplot(x='Atividade', y='Horas', data=dados)
plt.title('Distribuição de Esforço no Projeto')
plt.show()
```
